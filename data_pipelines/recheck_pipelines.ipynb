```python
import sys
from pathlib import Path

PROJECT_DIR = Path.cwd().parent
if str(PROJECT_DIR) not in sys.path:
    sys.path.insert(0, str(PROJECT_DIR))

from data_pipelines.labelling.step04_postprocess import _get_records, _get_events_fields, _get_field_diff

SCHEMA_PATH   = PROJECT_DIR / "data_pipelines/labelling/data_schema.json"
EVENTS_FIELDS = _get_events_fields(SCHEMA_PATH)
```

```python
def _get_output_error(output: list[dict] | None) -> list[str]:
    """
    - Summary: Liệt kê lỗi field của từng event trong output đã parse.
    - Args:
        - output: List event đã parse từ label_raw, None nếu parse lỗi.
    - Output:
        - list[str]: List mô tả lỗi field từng event, rỗng nếu output hợp lệ.
    """
    if output is None:
        return ["parse lỗi: không parse được label_raw"]

    errors: list[str] = []
    for event in output:
        event_type = event.get("event_type")
        if event_type not in EVENTS_FIELDS:
            errors.append(f"event_type lạ: '{event_type}'")
            continue
        missing_fields, extra_fields = _get_field_diff(event, EVENTS_FIELDS)
        if missing_fields or extra_fields:
            errors.append(f"event '{event_type}': thiếu {missing_fields or '{}'}, thừa {extra_fields or '{}'}")
    return errors
```
